```python
from verycloudy.config import DATA_DIR, FILEPATH_CLOUDS_CSV, FILEPATH_CLOUDS_APPEND_CSV, FILEPATH_CLOUDS_JSON, FILEPATH_CLOUDS_APPEND_JSON, FILEPATH_CLOUDS_TEST, FILEPATH_DATE_TEST, FILEPATH_DATE_CSV
```

```python
import pandas as pd
```

```python
df1 = pd.read_csv(FILEPATH_CLOUDS_CSV)
df2 = pd.read_csv(FILEPATH_CLOUDS_APPEND_CSV)

combined = pd.concat([df1, df2], ignore_index=True)

combined.to_csv(DATA_DIR / "combined.csv", index=False)
```

```python
with open(DATA_DIR / "combined.jsonl", "w", encoding="utf-8") as out:
    for fname in [FILEPATH_CLOUDS_JSON, FILEPATH_CLOUDS_APPEND_JSON]:
        with open(fname, "r", encoding="utf-8") as f:
            for line in f:
                if line.strip():
                    out.write(line)
```
